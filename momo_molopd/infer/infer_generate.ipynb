# MolOPD inference and JSON export

```python
import json
from transformers import AutoTokenizer, AutoModelForCausalLM
from peft import PeftModel
model_name = 'chandar-lab/NovoMolGen_32M_SMILES_AtomWise'
adapter_path = 'pretrained_mol/teachers/fda_lora'
tokenizer = AutoTokenizer.from_pretrained(model_name)
base = AutoModelForCausalLM.from_pretrained(model_name)
model = PeftModel.from_pretrained(base, adapter_path)
prompt = 'C'
inputs = tokenizer(prompt, return_tensors='pt')
out = model.generate(**inputs, max_new_tokens=64, do_sample=True, top_p=0.95, temperature=0.8, num_return_sequences=32)
smiles = tokenizer.batch_decode(out, skip_special_tokens=True)
payload = [{'id': i, 'smiles': s.strip()} for i, s in enumerate(smiles)]
with open('momo_molopd/infer/generated_molecules.json', 'w', encoding='utf-8') as f:
    json.dump(payload, f, ensure_ascii=False, indent=2)
print('saved', len(payload))
```
