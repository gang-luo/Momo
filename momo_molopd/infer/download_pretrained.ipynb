# Download NovoMolGen checkpoints and datasets to pretrained_mol/

```python
from huggingface_hub import snapshot_download
from pathlib import Path
root = Path('pretrained_mol')
root.mkdir(exist_ok=True, parents=True)
repos = [
 'chandar-lab/NovoMolGen_32M_SMILES_AtomWise',
 'chandar-lab/NovoMolGen_157M_SMILES_AtomWise',
 'chandar-lab/NovoMolGen_300M_SMILES_AtomWise',
 'chandar-lab/ZINC_250k',
 'chandar-lab/ZINC_22',
 'chandar-lab/zinc_scaffold',
 'chandar-lab/zinc_random',
]
for repo in repos:
    local_dir = root / repo.split('/')[-1]
    snapshot_download(repo_id=repo, repo_type='model' if 'NovoMolGen' in repo else 'dataset', local_dir=str(local_dir), local_dir_use_symlinks=False)
    print('downloaded', repo, '->', local_dir)
```
